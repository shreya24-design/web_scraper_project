```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://news.ycombinator.com/"
response = requests.get(url)
print("Website Connected!")

soup = BeautifulSoup(response.text, "html.parser")

headlines=[]

for item in soup.find_all("span", class_="titleline"):
  headlines.append(item.text)


  df = pd.DataFrame(headlines, columns=["Headlines"])

  print(df.head())

  df.to_csv("headlines.csv", index=False)

  print("CSV File Created Successfully!")
```

```python
from datetime import datetime

current_time = datetime.now()
print("Project run Date and Time:")
print(current_time)
```

```python
df["Scraped_Date"] = current_time.strftime("%Y-%m-%d")
df["Scraped_Time"] = current_time.strftime("%H:%M:%S")

print(df.head())
```

```python
df.to_csv("final_web_scraper_output.csv", index=False)

print("Final Project Saved Successfully!")
```
